```python
index_commands=[
    "CREATE INDEX IF NOT EXISTS idx_orders_customer_id ON Orders(customer_id);",
    "CREATE INDEX IF NOT EXISTS idx_orderitems_order_id ON OrderItems(order_id);",
    "CREATE INDEX IF NOT EXISTS idx_orderitems_product_id ON OrderItems(product_id);",
    "CREATE INDEX IF NOT EXISTS idx_payments_order_id ON Payments(order_id);",
    "CREATE INDEX IF NOT EXISTS idx_products_product_id ON Products(product_id);"
]

try:
    cursor=conn.cursor()
    for command in index_commands:
        cursor.execute(command)
    conn.commit()
    print("indexes created succesfully")
except Exception as e:
    print(f"error during index creation: {e}")

full_query = """
SELECT
    C.customer_city AS city,
    C.customer_state AS state,
    O.order_status AS order_status,
    O.order_purchase_timestamp AS order_date,
    OI.order_id,
    OI.product_id,
    OI.seller_id,
    OI.item_price,
    OI.shipping_charges,
    OI.item_revenue,
    P.payment_type,
    PR.product_category_name
FROM Orders O
JOIN Customers C
    ON O.customer_id = C.customer_id
JOIN (
    SELECT
        order_id,
        product_id,
        seller_id,
        SUM(price) AS item_price,
        SUM(shipping_charges) AS shipping_charges,
        SUM(price + shipping_charges) AS item_revenue
    FROM OrderItems
    GROUP BY order_id, product_id, seller_id
) OI
    ON O.order_id = OI.order_id
LEFT JOIN (
    SELECT
        order_id,
        MIN(payment_type) AS payment_type
    FROM Payments
    GROUP BY order_id
) P
    ON O.order_id = P.order_id
LEFT JOIN Products PR
    ON OI.product_id = PR.product_id;
"""
```

```python
import time
start=time.time()
new_table=pd.read_sql_query(full_query,conn)
end=time.time()
print(f"query executed in:{(end-start)/60} minutes")
```

```python
print("Shape:", new_table.shape)
print("\nColumns:")
print(new_table.columns.tolist())

display(new_table.head())
```

```python

print("Dataset shape:", new_table.shape)

print("\nDuplicate rows:", new_table.duplicated().sum())

print("\nMissing values:")
print(new_table.isnull().sum())

print("\nUnique orders:", new_table["order_id"].nunique())

print("\nUnique products:", new_table["product_id"].nunique())

print("\nRevenue totals:")
print("Item price:", round(new_table["item_price"].sum(), 2))
print("Shipping charges:", round(new_table["shipping_charges"].sum(), 2))
print("Item revenue:", round(new_table["item_revenue"].sum(), 2))
```

```python
print("Item price nulls:", new_table["item_price"].isnull().sum())
print("Shipping charges nulls:", new_table["shipping_charges"].isnull().sum())
print("Item revenue nulls:", new_table["item_revenue"].isnull().sum())

print(
    "Revenue calculation mismatches:",
    (
        (new_table["item_price"] + new_table["shipping_charges"])
        != new_table["item_revenue"]
    ).sum()
)

payment_counts = (
    new_table.groupby("order_id")["payment_type"]
    .nunique()
)

print("\nOrders with multiple payment types:", (payment_counts > 1).sum())
print("Orders with one payment type:", (payment_counts == 1).sum())
```

```python
missing_category = new_table[
    new_table["product_category_name"].isnull()
]

print("Rows with missing categories:", len(missing_category))

print(
    "Unique products with missing categories:",
    missing_category["product_id"].nunique()
)

print("\nSample missing-category products:")
print(
    missing_category[
        ["product_id", "city", "state", "item_price", "item_revenue"]
    ].head(10)
)

print(
    "\nAre these product IDs present in Products?",
    missing_category["product_id"].isin(
        pd.read_sql_query("SELECT product_id FROM Products", conn)["product_id"]
    ).all()
)
```

```python
missing_product_ids = missing_category["product_id"].unique().tolist()

placeholders = ",".join(["?"] * len(missing_product_ids))

category_check = pd.read_sql_query(
    f"""
    SELECT product_id, product_category_name
    FROM Products
    WHERE product_id IN ({placeholders})
    """,
    conn,
    params=missing_product_ids
)

print("Matching product records:", len(category_check))
print("\nCategory value counts:")
print(
    category_check["product_category_name"]
    .fillna("<NULL>")
    .value_counts(dropna=False)
)

print("\nSample source records:")
print(category_check.head(15))
```

```python
sales_clean = new_table.copy()

sales_clean["product_category_name"] = (
    sales_clean["product_category_name"]
    .fillna("Unknown")
)

print("Original rows:", len(new_table))
print("Cleaned rows:", len(sales_clean))

print(
    "\nMissing categories after cleaning:",
    sales_clean["product_category_name"].isnull().sum()
)

print("\nCategory distribution:")
print(
    sales_clean["product_category_name"]
    .value_counts()
    .head(10)
)

print(
    "\nUnknown category rows:",
    (sales_clean["product_category_name"] == "Unknown").sum()
)
```
